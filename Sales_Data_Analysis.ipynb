## Dataset Description

The dataset contains sales transaction records with 120 rows and 9 columns.

Each record provides information about an order, including its date, customer, city, product, category, unit price, quantity, and payment method.

**Columns in the dataset:**
- `order_id` – Unique order identifier
- `order_date` – Date of the transaction
- `customer_id` – Customer identifier
- `city` – City where the transaction occurred
- `product` – Product purchased
- `category` – Product category
- `unit_price` – Price per unit
- `quantity` – Number of units purchased
- `payment_method` – Payment method used

## Business Objectives

The main objectives of this analysis are:

1. Analyze overall sales revenue and transaction performance.
2. Identify the best-performing products and categories.
3. Evaluate sales performance across different cities.
4. Analyze monthly revenue trends and growth patterns.
5. Understand customer purchasing behaviour and segment customers based on revenue.
6. Identify repeat customers and evaluate their contribution to total revenue.
7. Generate actionable business insights from the sales data.

## Data Cleaning and Preparation

The following data cleaning and preparation steps were performed:

1. Loaded the sales dataset using Pandas.
2. Inspected the dataset structure, column data types, and summary statistics.
3. Identified missing values in the `city`, `unit_price`, and `quantity` columns.
4. Replaced the missing city value with "Unknown".
5. Checked for duplicate records and investigated repeated order IDs to distinguish valid transactions from duplicates.
6. Standardized text values by removing leading and trailing spaces and applying consistent capitalization.
7. Converted the `order_date` column into a datetime format.
8. Excluded records with missing unit prices or quantities from revenue calculations.
9. Created a new `revenue` column by multiplying unit price by quantity.
10. Extracted the month from the order date for monthly sales analysis.

## Key Findings

1. **City Performance:** Mumbai generated the highest total revenue (885,770), while Delhi recorded the highest quantity sold (77 units).

2. **Product Performance:** Laptop generated the highest product revenue (1,302,600), whereas Office Chair recorded the highest quantity sold (46 units).

3. **Category Performance:** Electronics was the leading category, generating 2,052,937 in revenue.

4. **Monthly Performance:** July recorded the highest monthly revenue (617,613). August experienced the largest month-over-month revenue decline.

5. **Customer Segmentation:** High-value customers contributed the largest share of total customer revenue.

6. **Customer Retention:** 29 out of 33 customers were repeat customers, representing approximately 87.88% of the customer base.

7. **Repeat Customer Contribution:** Repeat customers generated approximately 99.43% of total customer revenue.

## Business Recommendations

1. Focus on retaining high-value and repeat customers through loyalty programs and personalized offers.
2. Investigate the factors behind Mumbai's strong revenue performance and explore opportunities to replicate them in other cities.
3. Maintain sufficient inventory of high-revenue products such as laptops.
4. Investigate the significant monthly revenue fluctuations to identify possible seasonal patterns or sales opportunities.
5. Develop strategies to increase purchases from one-time and low-value customers.

```python
import pandas as pd

df = pd.read_csv("sales_analysis_dataset.csv")
df.head()
```

```python
df["city"] = df["city"].fillna("unknown")
```

```python
df.isnull().sum()
```

```python
sales_df = df[(df["unit_price"].notnull()&df["quantity"].notnull())].copy()
```

```python
sales_df
```

```python
sales_df.duplicated().sum()
```

```python
sales_df["order_id"].duplicated().sum()
```

```python
sales_df[sales_df["order_id"].duplicated(keep=False)]
```

```python
x = sales_df.groupby("order_id").agg(record_count=("order_id","count")).reset_index()
x[x["record_count"]>1]
```

```python
columns=["city","product","category"]
sales_df[columns] = sales_df[columns].apply(lambda x : x.str.strip().str.title())
```

```python
sales_df["payment_method"] = sales_df["payment_method"].str.strip().str.title()
```

```python
sales_df["order_date"] = pd.to_datetime(sales_df["order_date"],format="mixed")
```

```python
sales_df.info()
```

```python
sales_df["revenue"] = sales_df["unit_price"]*sales_df["quantity"]
```

```python
sales_df.head()
```

```python
summary = pd.DataFrame({"Metrics": ["total revenue","total quantity sold","average transaction revenue","total transactions records"],"value":[sales_df["revenue"],sales_df["quantity"].sum(),sales_df["revenue"].mean(),sales_df.shape]})
```

```python
summary
```

```python
sales_df.groupby("category").agg(total_quantity_sold=("quantity","sum"),
total_revenue=("revenue","sum"),average_transaction_revenue=("revenue","mean")).reset_index()
```

Interpretation
- Electronics is the best-performing category in both total quantity sold and total revenue.
- Furniture is the second-highest revenue-generating category.
- Stationery generates the lowest revenue, despite selling a similar quantity to Accessories.
- This shows an important business insight: higher sales quantity doesn't always mean higher revenue. Product prices also have a major impact.

```python
sales_df.groupby("product").agg(total_quantity_sold=("quantity","sum"),total_revenue=("revenue","sum")).sort_values(by="total_revenue",ascending=False).reset_index()
```

What we can learn
- Laptop is the top revenue-generating product, despite selling fewer units than some other products.
- Office Chair has the highest quantity sold (46 units), but generates much less revenue than Laptops and Smartphones.
- This again shows that sales volume and revenue are different metrics. A product can sell more units but generate less revenue if its unit price is lower.

```python
sales_df.groupby("city").agg(total_quantity_sold=("quantity","sum"),total_revenue=("revenue","sum"),average_transaction_revenue=("revenue","mean")).sort_values(by="total_revenue",ascending=False).reset_index()
```

Business observations
1. Mumbai generates the highest revenue, even though Delhi sells more units.
2. Delhi has the highest quantity sold among the identified cities.
3. Pune ranks third in total revenue, despite having relatively low sales volume.
4. The Unknown city category accounts for two units. We retained these records rather than deleting them.

```python
sales_df.groupby("payment_method").agg(total_transactions=("order_id","count"),total_revenue=("revenue","sum"),average_transactions_revenue=("revenue","mean")).sort_values(by="total_revenue",ascending=False).reset_index()
```

Business insights
1. Net Banking generates the highest revenue, narrowly ahead of Cash.
2. Credit Card has the highest transaction count (34), but doesn't generate the highest revenue.
3. UPI has the lowest total revenue and average transaction revenue.
4. Higher transaction counts don't necessarily translate into higher revenue.

```python
sales_df["month"] = sales_df["order_date"].dt.month
x = sales_df.groupby("month").agg(total_revenue=("revenue","sum"),total_quantity_sold=("quantity","sum"),total_transactions=("order_id","count")).sort_values("month").reset_index()
```

Business insights
- July had the highest revenue: 617,613.
- November was the second-highest month: 331,892.
- April had the lowest revenue: 80,145.
- December recorded the highest quantity sold: 38 units.
- August and October both sold 37 units.
- July generated the highest revenue with 34 units sold, suggesting a higher average selling price or a more premium product mix.
One important observation
December had the highest quantity sold, but July generated much higher revenue. This shows why businesses should examine both sales volume and revenue.

```python
x = sales_df.groupby("month").agg(
    total_revenue=("revenue", "sum"),
    total_quantity_sold=("quantity", "sum"),
    total_transactions=("order_id", "count")
).sort_values("month").reset_index()

x.columns
x["previous_month_revenue"] = x["total_revenue"].shift(1)
x["revenue_change"] = x["total_revenue"] - x["previous_month_revenue"]
```

```python
x["revenue_growth_percentage"] = x["revenue_change"]/x["previous_month_revenue"]*100
```

```python
customer_summary = sales_df.groupby("customer_id").agg(total_revenue=("revenue","sum"),total_quantity_purchased=("quantity","sum"),total_transactions=("order_id","count")).sort_values(by="total_revenue",ascending=False).reset_index()
```

```python
customer_summary.loc[customer_summary["total_revenue"] > 50000, "customer_segment"] = "High"
```

```python
customer_summary.loc[customer_summary["total_revenue"].between(20000, 50000), "customer_segment"] = "Medium"
```

```python
customer_summary.loc[customer_summary["total_revenue"] < 20000, "customer_segment"] = "Low"
```

```python
customer_summary
```

Business insights
1. 17 customers (51.52%) belong to the High Value segment.
2. 6 customers (18.18%) belong to the Medium Value segment.
3. 10 customers (30.30%) belong to the Low Value segment.
Business interpretation: More than half of the customers generate over ₹50,000 in revenue. 
The business should focus on retaining these high-value customers while developing strategies to increase spending among medium- and low-value customers.
